# EnterpriseKG synthetic data loader
Generate fresh sample data with `_gen_demo_data.py`, then upload its JSON files into `Files/demo` in your own development lakehouse.
This notebook overwrites ENT_* demo tables. No source customer payloads are included.

```python
from pyspark.sql.functions import col
names = ['ENT_Staff', 'ENT_Capability', 'ENT_Skill', 'ENT_StaffCapability', 'ENT_StaffSkill', 'ENT_PracticeNetworkMembers', 'ENT_Client', 'ENT_Programme', 'ENT_Project', 'ENT_ProjectStaff', 'ENT_ProjectFinancialCumulative', 'ENT_ProjectDisciplineStatistics']
for name in names:
    target = spark.table(name.lower())
    frame = spark.read.option('multiLine', True).json(f'Files/demo/{name}.json')
    for field in target.schema.fields:
        frame = frame.withColumn(field.name, col(field.name).cast(field.dataType))
    frame.select(*target.columns).write.mode('overwrite').format('delta').saveAsTable(name.lower())
    print(name, spark.table(name.lower()).count())
```
