# EnterpriseKG Lakehouse — table bootstrap

Creates the 12 ENT_* Delta tables in `EnterpriseKG_Lakehouse`.

Run after the workspace + lakehouse exist, before loading demo data.

```python
spark.sql('''CREATE TABLE IF NOT EXISTS ent_staff (
  `StaffID` INT,
  `StaffUPN` STRING,
  `StaffName` STRING,
  `StaffStatus` STRING,
  `Email` STRING,
  `OfficeCode` STRING,
  `OfficeName` STRING,
  `CountryName` STRING,
  `RegionName` STRING,
  `PracticeName` STRING,
  `DisciplineName` STRING,
  `Grade` STRING,
  `JobTitle` STRING,
  `StartDate` DATE,
  `LeaveDate` DATE,
  `ManagerID` INT,
  `ManagerName` STRING,
  `ManagerEmail` STRING,
  `ReportsToUPN` STRING,
  `MyPeoplePageURL` STRING
) USING DELTA''')
```

```python
spark.sql('''CREATE TABLE IF NOT EXISTS ent_capability (
  `CapabilityID` BIGINT,
  `CapabilityCode` STRING,
  `CapabilityName` STRING,
  `MarketName` STRING,
  `TechnicalArea` STRING,
  `StrategicTheme` STRING,
  `Billable` BOOLEAN
) USING DELTA''')
```

```python
spark.sql('''CREATE TABLE IF NOT EXISTS ent_skill (
  `SkillID` BIGINT,
  `SkillCode` STRING,
  `SkillName` STRING,
  `SkillCategory` STRING,
  `RelatedCapabilityCode` STRING,
  `DigitalSkill` BOOLEAN
) USING DELTA''')
```

```python
spark.sql('''CREATE TABLE IF NOT EXISTS ent_staffcapability (
  `StaffID` INT,
  `StaffName` STRING,
  `StaffUPN` STRING,
  `CapabilityCode` STRING,
  `CapabilityName` STRING,
  `ProficiencyLevel` STRING,
  `YearsExperience` INT,
  `Accredited` BOOLEAN,
  `PrimaryCapability` BOOLEAN
) USING DELTA''')
```

```python
spark.sql('''CREATE TABLE IF NOT EXISTS ent_staffskill (
  `StaffID` INT,
  `StaffName` STRING,
  `StaffUPN` STRING,
  `SkillCode` STRING,
  `SkillName` STRING,
  `ProficiencyLevel` STRING,
  `YearsExperience` INT,
  `PrimarySkill` BOOLEAN
) USING DELTA''')
```

```python
spark.sql('''CREATE TABLE IF NOT EXISTS ent_practicenetworkmembers (
  `PracticeNetworkID` BIGINT,
  `PracticeNetworkName` STRING,
  `StaffID` INT,
  `StaffEmail` STRING,
  `StaffUPN` STRING,
  `StaffName` STRING,
  `Current` BOOLEAN
) USING DELTA''')
```

```python
spark.sql('''CREATE TABLE IF NOT EXISTS ent_client (
  `ClientID` INT,
  `ClientCode` STRING,
  `ClientName` STRING,
  `ClientStatusDesc` STRING,
  `ClientSectorName` STRING,
  `CountryName` STRING,
  `RelationshipDirectorID` INT,
  `RelationshipDirectorName` STRING,
  `RelationshipDirectorEmail` STRING,
  `StrategicClient` BOOLEAN,
  `FrameworkClient` BOOLEAN,
  `ActiveProjectCount` INT,
  `TotalBacklogValue` DECIMAL(38,18)
) USING DELTA''')
```

```python
spark.sql('''CREATE TABLE IF NOT EXISTS ent_programme (
  `ProgrammeID` INT,
  `ProgrammeCode` STRING,
  `ProgrammeName` STRING,
  `ClientID` INT,
  `ClientName` STRING,
  `MarketName` STRING,
  `RegionName` STRING,
  `StartDate` DATE,
  `EndDate` DATE,
  `ProgrammeDirectorID` INT,
  `ProgrammeDirectorName` STRING,
  `Status` STRING,
  `TotalProgrammeValue` DECIMAL(38,18)
) USING DELTA''')
```

```python
spark.sql('''CREATE TABLE IF NOT EXISTS ent_project (
  `ProjectID` INT,
  `ProjectCode` STRING,
  `ShortTitle` STRING,
  `FullTitle` STRING,
  `ClientID` INT,
  `ClientName` STRING,
  `ProgrammeID` INT,
  `ProgrammeName` STRING,
  `MarketName` STRING,
  `SectorName` STRING,
  `CountryName` STRING,
  `RegionName` STRING,
  `ProjectStatusDesc` STRING,
  `StartDate` DATE,
  `EndDate` DATE,
  `ProjectDirectorID` INT,
  `ProjectDirectorName` STRING,
  `ProjectManagerID` INT,
  `ProjectManagerName` STRING,
  `CommercialManagerID` INT,
  `CommercialManagerName` STRING,
  `FrameworkName` STRING,
  `ContractModel` STRING,
  `TotalFee` DECIMAL(38,18),
  `PercentComplete` DECIMAL(38,18),
  `BIMEnabled` BOOLEAN,
  `DigitalTwinEnabled` BOOLEAN
) USING DELTA''')
```

```python
spark.sql('''CREATE TABLE IF NOT EXISTS ent_projectstaff (
  `ProjectID` INT,
  `ProjectCode` STRING,
  `StaffID` INT,
  `StaffName` STRING,
  `StaffUPN` STRING,
  `RoleName` STRING,
  `RoleType` STRING,
  `DisciplineName` STRING,
  `HoursBooked` DECIMAL(34,6),
  `ChargeabilityPct` DECIMAL(10,2),
  `StartDate` DATE,
  `EndDate` DATE,
  `Current` BOOLEAN
) USING DELTA''')
```
